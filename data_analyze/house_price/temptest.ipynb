```python
soup = bs4.BeautifulSoup(html, 'html.parser')
```

```python
pattern = r'(\d{4})年(\d{1,2})月份70个大中城市商品住宅销售价格变动情况'


def match_title(s):
    match = re.fullmatch(pattern, s)
    year = ""
    month = ""
    if match:
        year = match.group(1)
        month = match.group(2).zfill(2)
        # print(f"匹配成功: '{s}' → 年份: {year}, 月份: {month}")
        return year+month
    else:
        # print(f"不匹配: '{s}'")
        pass
        return None
    
    
```

```python
for i in href:
    res = match_title(i.text)
    if res:
        print(res)
    # print(i.text)
```

```python
city_cols = [col for col in df.columns if col != '时间']
startdate = 201712
df2=df[df["时间"]>startdate]
# 新建结果 DataFrame，第一行全是 100
result = pd.DataFrame(columns=df2.columns)
first_row = [str(startdate)] + [100.0] * len(city_cols)
result.loc[0] = first_row

# 依次计算后续行
for i, row in df2.iterrows():
    prev = result.iloc[-1, 1:].astype(float)  # 上一行的城市数据
    curr = row[city_cols].astype(float)      # 当前行的指数
    new_row = [row['时间']] + list(prev.values * curr.values / 100)
    result.loc[len(result)] = new_row

# 重置索引
result = result.reset_index(drop=True)
result.to_csv("processed.csv",index=False, encoding='utf-8-sig')
# print(result)
```
